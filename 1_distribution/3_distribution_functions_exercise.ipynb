## Case Study: Analyzing Exam Scores with the Normal Distribution

You are given a list of exam scores representing students’ performance in a class. Solve the following tasks to analyze the distribution of these scores using normal distribution concepts. The list of exam scores is:

```python
exam_scores = [
    85,
    90,
    78,
    92,
    88,
    95,
    70,
    60,
    82,
    75,
    80,
    98,
    88,
    91,
    83,
    77,
    89,
    94,
    86,
    72,
]
```

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import norm
```

#### 1. Calculate the mean and standard deviation of the scores.
- Hint: Use the numpy mean and std functions.

```python
mean_score = np.mean(exam_scores)
std_score = np.std(exam_scores)

print(f"Mean score: {mean_score:.2f}")
print(f"Standard deviation: {std_score:.2f}")
```

#### 2. Plot the distribution of the score as a histogram.


```python
sns.histplot(exam_scores, color="#ff4a11ff")
plt.title("Exam Scores")
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.show()
```

#### 3. Plot the normal distribution curve of the scores.

- Hint: To visualize the normal distribution curve, you can use the norm.pdf function from scipy.stats (more details [here](https://www.mathsisfun.com/data/random-variables.html)) You will need to specify the mean and standard deviation calculated in Task 1 and create spaced numbers over the range of the scores which can be done using:

```python

x_range = np.linspace(min(exam_scores), max(exam_scores), 100)
pdf = norm.pdf(x_range, mean_score, std_score)

plt.plot(x_range, pdf, color="#ff4a11ff")
plt.title("Normal Curve Fitted to Exam Scores")
plt.xlabel("Score")
plt.ylabel("Probability Density")
plt.show()
```

4. Calculate the percentage of students who scored above a certain threshold score (for example above 80)

    - Hint: In this case, you can use norm.cdf from scipy.stats

(more details [here](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.norm.html)) as the following:

```python
scores = np.array(exam_scores)

model = 1 - norm.cdf(80, mean_score, std_score)
reality = np.mean(scores > 80)

print(f"model:   {model:.3f}")
print(f"reality: {reality:.3f}")
```

**Result:** Mean score 83.65, std 9.30. The histogram is left-skewed (one low score of 60, ceiling at 100), so the bell fits only roughly: its peak sits on the mean (~84) while the real peak is 85-90, and it predicts 3.9% of students above 100, which is impossible.
Share above 80: model 0.653 vs reality 0.650 (13 of 20). The bell is reliable near the centre but not in the tails (above 95: model 11.1% vs reality 5%).

## Case Study: Examining the Heights of Employees in a Company

A human resources department wants to understand the distribution of employee heights in their company. They suspect that employee heights follow a normal distribution and want to use this information to make decisions about uniform sizes, ergonomic furniture, and health and wellness programs.


Hint: Follow the steps in the overview and also use the table above.

## Data Generation

```python
# Do not alter this code
# Generating synthetic data for employee heights
np.random.seed(42)
data = np.random.normal(
    loc=170, scale=10, size=500
)  # Mean height of 170 cm, std deviation of 10 cm
```

## Your playground

```python
mean_height = np.mean(data)
std_height = np.std(data)

print(f"Mean height: {mean_height:.2f} cm")
print(f"Standard deviation: {std_height:.2f} cm")
```

```python
sns.histplot(data, color="#ff4a11ff")
plt.title("Heights")
plt.xlabel("Height (cm)")
plt.ylabel("Frequency")
plt.show()
```

```python
x_range = np.linspace(min(data), max(data), 100)
pdf = norm.pdf(x_range, mean_height, std_height)

sns.histplot(data, stat="density", color="#ff4a11ff")
plt.plot(x_range, pdf, color="black")
plt.title("Normal Curve Fitted to Heights")
plt.xlabel("Height (cm)")
plt.ylabel("Probability Density")
plt.show()
```

```python
heights = np.array(data)

model = 1 - norm.cdf(190, mean_height, std_height)
reality = np.mean(heights > 190)

print(f"model:   {model:.3f}")
print(f"reality: {reality:.3f}")
```

**Result:** 500 heights, mean 170.07 cm, std 9.80 cm. The histogram is symmetric and the fitted normal curve sits on the bars (density overlay), so a normal model fits well, also in the tails.
Business question (HR): share of employees taller than 190 cm who need a special tall uniform size: model 0.021 vs reality 0.026 (13 of 500). Planning rule: about 2-3% of staff.

## Case Study: Call Center Call Volume

Scenario: A call center wants to analyze the number of calls received per hour during its busiest periods. They suspect that the number of calls per hour follows a Poisson distribution. 


Hint: Follow the steps in the overview and also use the tables above

## Data Generation

```python
# Do not alter this code
from scipy.stats import poisson

# Generating synthetic data for the number of calls per hour
np.random.seed(2)
lambda_ = 8  # Average number of calls per hour
data = np.random.poisson(lam=lambda_, size=100)  # Simulated call volumes
```

## Your playground

```python
mean_calls = np.mean(data)
variance_calls = np.var(data)

print(f"Mean calls per hour: {mean_calls:.2f}")
print(f"Variance: {variance_calls:.2f}")
```

```python
x_values = np.arange(data.min(), data.max() + 1)
pmf = poisson.pmf(x_values, mean_calls)

sns.histplot(data, discrete=True, stat="probability", color="#ff4a11ff")
plt.plot(x_values, pmf, "o-", color="black")
plt.title("Poisson PMF Fitted to Calls per Hour")
plt.xlabel("Calls per hour")
plt.ylabel("Probability")
plt.show()
```

```python
model = poisson.sf(11, mean_calls)
reality = np.mean(data >= 12)

print(f"model:   {model:.3f}")
print(f"reality: {reality:.3f}")
```

**Result:** 100 hours, mean 7.90 calls, variance 8.13. Mean ≈ variance (Poisson fingerprint) and the Poisson PMF follows the bars, so a Poisson model fits; bars are jagged because 100 hours is a small sample.
Business question: the team can handle 11 calls per hour. Share of hours with 12 or more calls: model 0.105 (poisson.sf(11)) vs reality 0.120. Roughly 1 hour in 10 is overloaded, which supports adding extra staff in busy hours.
